#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.

```r
# Calculate the mean, minimum, and maximum for all numeric columns, dropping missing values
numeric_summary <- march_madness_tidy |>
  summarize(
    across(
      where(is.numeric),
      list(
        mean = ~ mean(.x, na.rm = TRUE),
        min = ~ min(.x, na.rm = TRUE),
        max = ~ max(.x, na.rm = TRUE)
      )
    )
  )

# Show the output table
numeric_summary
```

```r
# Plot 1: Distribution of a numeric variable (Regional Win Percentage)
options(repr.plot.width = 8, repr.plot.height = 5)

reg_win_dist_plot <- march_madness_tidy |>
  drop_na(reg_wins_pct) |>
  ggplot(aes(x = reg_wins_pct)) +
  geom_histogram(bins = 25, fill = "steelblue", color = "white") +
  labs(
    title = "Distribution of Regular-Season Regional Win Percentage",
    x = "Regional Win Percentage (%)",
    y = "Number of Teams"
  ) +
  theme_minimal(base_size = 14)

reg_win_dist_plot
```

```r
# Plot 2: Relative proportions of each category in the categorical variable (Bid Type)
bid_prop_plot <- march_madness_tidy |>
  drop_na(bid) |>
  count(bid) |>
  mutate(proportion = n / sum(n)) |>
  ggplot(aes(x = bid, y = proportion, fill = bid)) +
  geom_bar(stat = "identity", width = 0.6) +
  labs(
    title = "Relative Proportions of Tournament Bid Types",
    x = "Tournament Qualification Bid Type",
    y = "Relative Proportion of Teams",
    fill = "Bid Type"
  ) +
  theme_minimal(base_size = 14)

bid_prop_plot
```

Visualization 1: This boxplot compares the distribution of conference win percentages (conf_wins_pct) between teams that earned an automatic bid versus an at-large bid. Because automatic bids are awarded to conference champions, we expect automatic-bid teams to have higher conference win percentages than at-large teams. Seeing a clear difference between the two groups confirms that conference win percentage is an informative predictor for classifying a team's tournament bid type.

```r
# Visualization 1: Conference Win Percentage by Tournament Bid Type
conf_vs_bid_plot <- march_madness_tidy |>
  drop_na(bid, conf_wins_pct) |>
  ggplot(aes(x = bid, y = conf_wins_pct, fill = bid)) +
  geom_boxplot(alpha = 0.8) +
  labs(
    title = "Conference Win Percentage by Tournament Bid Type",
    x = "Tournament Qualification Bid Type",
    y = "Conference Win Percentage (%)",
    fill = "Bid Type"
  ) +
  theme_minimal(base_size = 14)

conf_vs_bid_plot
```

Visualization 2: This scatter plot explores the relationship between tournament seed, regular-season regional win percentage (reg_wins_pct), and qualification bid type (bid). At-large bids are typically awarded to strong teams from competitive conferences that did not win their conference title, while automatic bids include champions from lower-ranked conferences that receive higher-numbered seeds. Visualizing how bid types separate across seed and regional win percentage helps evaluate how well these predictors work together for classification.

```r
# Visualization 2: Regional Win Percentage vs Tournament Seed colored by Bid Type
seed_reg_scatter_plot <- march_madness_tidy |>
  drop_na(bid, seed, reg_wins_pct) |>
  ggplot(aes(x = seed, y = reg_wins_pct, color = bid)) +
  geom_point(alpha = 0.5, size = 2) +
  labs(
    title = "Regional Win Percentage vs. Tournament Seed by Bid Type",
    x = "Tournament Seed (1 = Highest, 16 = Lowest)",
    y = "Regional Win Percentage (%)",
    color = "Bid Type"
  ) +
  theme_minimal(base_size = 14)

seed_reg_scatter_plot
```
